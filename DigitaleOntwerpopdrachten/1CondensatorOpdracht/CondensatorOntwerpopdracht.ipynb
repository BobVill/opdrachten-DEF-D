4. Print het exponent op een nette manier.

````python
from scipy.optimize import curve_fit
###Hier jouw code###
#code to find index (1)
index_z0 = (np.abs(z)).argmin()
print('De index bij z=0 m is:',index_z0)
#otherwise
index_z0manual = int(height_domain / grid_step)
#print('De index bij z=0 m is:',index_z0manual)
###

# plot van de potentiaal tegen de straal en curve-fit (2+3)
Vz0 = V[index_z0]  # potentiaal op z=0 m
Vz0_zonder0=Vz0[1:]  # verwijder de potentiaal op r=0 m (bestaat niet)
r_zonder0=r[1:]  # verwijder de straal op r=0 m
from scipy.optimize import curve_fit
def log_potentiaal(logstraal, exponent, logcoefficient):
    return logcoefficient + exponent * logstraal
val, cov=curve_fit(log_potentiaal, np.log(r_zonder0), np.log(Vz0_zonder0))

plt.figure()
plt.plot(np.log(r_zonder0), np.log(Vz0_zonder0), 'o', label='datapunten')
plt.plot(np.log(r_zonder0), log_potentiaal(np.log(r_zonder0), *val), 'g',label='curve-fit')
plt.xlabel('$log(r)$ (log(m))')
plt.ylabel('$log(V)$ (log(V))')
plt.title('Logaritme van de potentiaal tegen de logaritme van de straal op z=0 m')
plt.legend()
plt.show()

# Exponent printen (4)
print('De exponent is:', val[0])
````

$$ \mathbf{E} = \frac{\sigma}{2\epsilon_0} \bigg[ 1- \frac{z}{\sqrt{R^2 + z^2}}\bigg] \hat{\mathbf{z}}$$
Zie probleem 2.6 in het boek van Griffith.

1. Wat moet de hoogte $z$ zijn boven de schijf zodat het elektrische veld door de schijf 10% is in vergelijking met dat van een oneindig geladen plaat?

Voeg hieronder je analytische uitwerking toe.

##### Antwoord 2h

Bij en oneindige plaat is R oneindig groot, dus de term $z/\sqrt{R^2 + z^2}$ gaat naar 0 en het veld is $\sigma/(2\epsilon_0)$.

$$E_{disk} = 0.10 \cdot E_{plate}$$
$$\frac{\sigma}{2\epsilon_0} \left[ 1 - \frac{z}{\sqrt{R^2 + z^2}} \right] = 0.10 \cdot \frac{\sigma}{2\epsilon_0}$$
$$1 - \frac{z}{\sqrt{R^2 + z^2}} = 0.10$$
$$\frac{z}{\sqrt{R^2 + z^2}} = 0.9$$
$$0.81 R^2 + 0.81 z^2 = z^2$$
$$z = \sqrt{\frac{81}{19}} R$$

````python
def vlakkeplaat(oppervlakte, afstand, epsilonR):
    C = epsilon0*epsilonR*oppervlakte/afstand 
    return C

def cilinder(lengte, straal_binnen, straal_buiten, epsilonR):
    C = 2*np.pi*epsilon0*epsilonR*lengte / np.log(straal_buiten/straal_binnen)
    return  C
````

````python
# Schrijf hier de code voor je oplossing. Gebruik hier je eerder gemaakte functies
C_meting2 = np.array([128.2, 111.5, 105.5, 91.9, 88.8, 69.0, 62.6, 53.1]) # in pF
M_meting2 = np.array([0, 0.111, 0.2744, 0.434, 0.482, 0.6425, 0.8325, 0.9725]) # in kg
#Fit zonder dielectric
def lineaire_fit(M, a, b):
    return a * M + b
# Uitvoeren van de fit
popt2, pcov2 = curve_fit(lineaire_fit, M_meting2, C_meting2)
a_fit2, b_fit2 = popt2
# Berekenen van de veerconstante k
g = 9.81 # m/s^2
epsilon_lucht = epsilon0

C_F = C_meting2 * 1e-12 # Capaciteit in Farad
d0 = epsilon0 * 0.01 / C_F[0] # aftsand tussen de platen berekenen
w = 0.1       # m (breedte plaat)
A = C_F * d0 / epsilon_lucht   # oppervlakte in m^2
u = (0.01 - A) / w   # uitrekking in m
# uitrekking per massa

F = M_meting2 * g # kracht in Newton
# Formule veerconstante
k_perpunt = F / u # de veerconstante per punt

# Resultaat printen)
k_mean = np.mean(k_perpunt[1:])   # gemiddelde, eerste punt (0 massa) overslaan

u_theorie = (M_meting2 * g) / k_mean  # m
C_theorie = epsilon_lucht * w * (0.1 - u_theorie) / d0  # F


# def condensator_extra_laag(#jouw variabelen): # Schrijf hier jouw functie
    ## je functie
epsilon_0 = 8.854e-12 
g = 9.81              
    

t = 0.0005            # Dikte van plastic laag (gok)
eps_r_layer = 6      # Relatieve permittiviteit van plastic laag (ook een gok)
s = 0.0005              # Afstand tussen de platen
eps_r_spacing = 1.0    # Relatieve permittiviteit lucht = 1.0


def bereken_capaciteit(m, k=k_mean, L0=0.1, W=0.1):

    delta_x = (m * g) / k
    
    #nieuwe overlap lengte (L)
    #np.maximum om te voorkomen dat de overlap negatief wordt als de massa te zwaar is
    L = np.maximum(L0 - delta_x, 0)
    
    #noemer van de breuk uit formule
    noemer = (2 * t / eps_r_layer) + (s / eps_r_spacing)
    
    #capaciteit berekenen
    C = (L * W * epsilon_0) / noemer
    
    return C

massa_array = np.linspace(0, 0.3, 100) 

# Bereken
capaciteit_array = bereken_capaciteit(M_meting2)

### Maak hier je plot
plt.plot(M_meting2, C_theorie*1e12, 'r')

plt.plot(M_meting2, capaciteit_array* 1e12,'r--', linewidth=2)
plt.plot(M_meting2, C_meting2, color='blue', linewidth=2)
plt.title('Analytische Kalibratiegrafiek: Capaciteit vs Massa')
plt.xlabel('Massa (Kg)')
plt.ylabel('Capaciteit (pF)')
plt.show()
````
